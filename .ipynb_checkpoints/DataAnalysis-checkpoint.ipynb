## 1. Importation des bibliothèques et chargement des données
Dans cette section, nous chargeons les bibliothèques d'analyse de données (`pandas`, `numpy`), de visualisation (`matplotlib`, `seaborn`) et les modules de `scikit-learn` pour le Machine Learning. Nous importons ensuite les données consolidées issues de notre pipeline dans un DataFrame pandas.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')
sns.set_theme(style="whitegrid")
```

```python
# Chargement des données d'Avis et d'Alertes depuis le backup
df_avis = pd.read_csv("../backup/full_cleaned_AVIS.csv")
df_alert = pd.read_csv("../backup/full_cleaned_ALERT.csv")

# Identification des types de bulletins
df_avis["Type"] = "Avis"
df_alert["Type"] = "Alerte"

# Combinaison des données
df = pd.concat([df_avis, df_alert], ignore_index=True)

# Nettoyage et renommage des colonnes pour correspondre aux attentes du notebook
def clean_numeric(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip()
    if val_str.lower() in ["non disponible", "nan", "", "none", "n/a"]:
        return np.nan
    try:
        return float(val_str)
    except:
        return np.nan

def get_base_severity(score):
    if pd.isna(score):
        return "Non disponible"
    try:
        score = float(score)
    except:
        return "Non disponible"
    if score == 0.0:
        return "NONE"
    elif 0.1 <= score <= 3.9:
        return "LOW"
    elif 4.0 <= score <= 6.9:
        return "MEDIUM"
    elif 7.0 <= score <= 8.9:
        return "HIGH"
    elif 9.0 <= score <= 10.0:
        return "CRITICAL"
    else:
        return "Non disponible"

# Renommer les colonnes existantes
df = df.rename(columns={
    "title": "Titre",
    "published": "Date publication",
    "cve_id": "CVE",
    "cvss_score": "Score CVSS",
    "cwe": "Type CWE",
    "epss_score": "Score EPSS",
    "vendor": "Vendor",
    "product": "Produit",
    "versions": "Versions affectées",
    "description_cve": "Description",
    "link": "Lien bulletin"
})

df["Score CVSS"] = df["Score CVSS"].apply(clean_numeric)
df["Score EPSS"] = df["Score EPSS"].apply(clean_numeric)
df["Base Severity"] = df["Score CVSS"].apply(get_base_severity)

df["Type CWE"] = df["Type CWE"].fillna("Non disponible").astype(str).str.strip().replace({"nan": "Non disponible", "": "Non disponible", "None": "Non disponible"})
df["Vendor"] = df["Vendor"].fillna("Non disponible").astype(str).str.strip().replace({"nan": "Non disponible", "": "Non disponible", "None": "Non disponible"})

# Tri par date de publication (décroissante) pour trier les données chronologiquement
df = df.sort_values(by="Date publication", ascending=False).reset_index(drop=True)

print(f"Nombre de lignes : {df.shape[0]}")
print(f"Nombre de colonnes : {df.shape[1]}")

# Afficher un échantillon représentatif contenant à la fois des Avis et des Alertes
pd.concat([df[df["Type"] == "Avis"].head(3), df[df["Type"] == "Alerte"].head(2)])
```

### 3.13. Versions de logiciels les plus fréquemment touchées
Nous analysons ici les versions de produits les plus touchées afin d'identifier si des versions spécifiques reviennent très régulièrement dans les failles recensées.

```python
import ast

def parse_versions(v):
    if pd.isna(v):
        return []
    v_str = str(v).strip()
    if not v_str:
        return []
    try:
        if v_str.startswith('[') and v_str.endswith(']'):
            parsed = ast.literal_eval(v_str)
            if isinstance(parsed, list):
                return parsed
        return [v_str]
    except:
        return [v_str]

df_versions = df.copy()
df_versions['versions_list'] = df_versions['Versions affectées'].apply(parse_versions)
all_versions = [item for sublist in df_versions['versions_list'] for item in sublist if str(item).strip() not in ['', 'nan', 'n/a', 'Non disponible']]
version_counts = pd.Series(all_versions).value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=version_counts.values, y=version_counts.index, palette='Blues_r')
plt.title('Top 10 des versions de logiciels les plus fréquemment affectées', fontsize=14)
plt.xlabel("Nombre d'''occurrences", fontsize=12)
plt.ylabel('Version', fontsize=12)
plt.savefig('top_affected_versions.png', dpi=100, bbox_inches='tight')
plt.show()
```

## 4. Machine Learning Non Supervisé (Clustering)
Nous cherchons à partitionner les vulnérabilités en plusieurs groupes (clusters) homogènes selon leur Score CVSS et leur Score EPSS. Nous utilisons l'algorithme K-Means. Nous appliquons d'abord la méthode du coude pour déterminer le nombre optimal de clusters.


```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Nettoyage pour le Clustering
df_ml = df.dropna(subset=['Score CVSS', 'Score EPSS'])
X_clust = df_ml[['Score CVSS', 'Score EPSS']]

# Standardisation des données
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_clust)

# Détermination du K optimal par la méthode du coude
wcss = []
k_range = range(1, 8)
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(k_range, wcss, 'o-', color='navy', linewidth=2)
plt.title("Méthode du coude pour le choix de K", fontsize=14)
plt.xlabel("Nombre de clusters (k)", fontsize=12)
plt.ylabel("Inertie (WCSS)", fontsize=12)
plt.savefig("elbow_method.png", dpi=100, bbox_inches='tight')
plt.show()
```

```python
# Choix de K = 3 clusters
kmeans_opt = KMeans(n_clusters=3, random_state=42, n_init=10)
df_ml['Cluster'] = kmeans_opt.fit_predict(X_scaled)

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_ml, x='Score CVSS', y='Score EPSS', hue='Cluster', palette='Set1', alpha=0.6)
plt.title("Clustering K-Means (CVSS vs EPSS) avec K=3", fontsize=14)
plt.xlabel("Score CVSS", fontsize=12)
plt.ylabel("Score EPSS", fontsize=12)
plt.savefig("kmeans_clusters.png", dpi=100, bbox_inches='tight')
plt.show()

print("=== Valeurs moyennes des indicateurs par cluster ===")
print(df_ml.groupby('Cluster')[['Score CVSS', 'Score EPSS']].mean())
```

## 5. Machine Learning Supervisé (Classification)
Notre but est de prédire si une vulnérabilité présente une **criticité extrême (Score CVSS >= 9.0, c'est-à-dire une classification 'CRITICAL')**. Nous utilisons pour cela les attributs de la vulnérabilité comme le score EPSS, le fait que ce soit une alerte ou un avis ANSSI, et les CWE / Vendors les plus représentés. Nous entraînons un classifieur de type **Random Forest**.


```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Variable cible
df_ml['Is_Critical'] = (df_ml['Score CVSS'] >= 9.0).astype(int)

# Features
df_ml['Is_Alert'] = (df_ml['Type'] == 'Alerte').astype(int)

# Encodage binaire pour les CWE et Vendors les plus courants
top_cwe_list = df_ml['Type CWE'].value_counts().head(10).index
top_vendor_list = df_ml['Vendor'].value_counts().head(10).index

X = df_ml[['Score EPSS', 'Is_Alert']].copy()
for cwe in top_cwe_list:
    X[f'CWE_{cwe}'] = (df_ml['Type CWE'] == cwe).astype(int)
for vendor in top_vendor_list:
    X[f'Vendor_{vendor}'] = (df_ml['Vendor'] == vendor).astype(int)

y = df_ml['Is_Critical']

# Séparation d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Entraînement
rf = RandomForestClassifier(n_estimators=100, random_state=42, max_depth=10)
rf.fit(X_train, y_train)

# Prédiction
y_pred = rf.predict(X_test)

# Évaluation
print("=== Performances du Modèle de Classification Random Forest ===")
print(f"Accuracy du modèle : {accuracy_score(y_test, y_pred):.4f}\n")
print("Rapport de classification :")
print(classification_report(y_test, y_pred))

# Matrice de confusion
plt.figure(figsize=(6, 5))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.title("Matrice de Confusion", fontsize=14)
plt.ylabel("Classe réelle", fontsize=12)
plt.xlabel("Classe prédite", fontsize=12)
plt.savefig("confusion_matrix.png", dpi=100, bbox_inches='tight')
plt.show()
```

```python
# Tracé de l'importance des variables
importances = rf.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(12, 6))
sns.barplot(x=importances[indices[:10]], y=X.columns[indices[:10]], palette="Blues_r")
plt.title("Importance des 10 variables prédictives les plus significatives", fontsize=14)
plt.xlabel("Importance", fontsize=12)
plt.ylabel("Variable", fontsize=12)
plt.savefig("feature_importances.png", dpi=100, bbox_inches='tight')
plt.show()
```
